```python
#part 0
import os
from google.colab import userdata
from openai import OpenAI

API_KEY = userdata.get('GROQ_API_KEY')
if not API_KEY:
  raise ValueError("GROQ_API_KEY not found in environment variables")
print('API key loaded successfully.')

client = OpenAI(
    api_key=API_KEY,
    base_url="https://api.groq.com/openai/v1"
)

MODEL = "llama-3.3-70b-versatile"
print("client ready")
```

```python
#part 1.1
def ask_llm(user_prompt, system_prompt = "You are a helpful assistant.",
  temperature = 0.7, max_tokens = 500):
  response = client.chat.completions.create(
    model=MODEL,
    messages=[
      {"role": "system", "content": system_prompt},
      {"role": "user", "content": user_prompt}
    ],
    temperature=temperature,
    max_tokens=max_tokens
  )
  return response

response = ask_llm("What is artificial intelligence?")
print(response.choices[0].message.content)

print("\nUsage:")
print(response.usage)
```

**Anatomy of a call**
1. The system role provides instructions that define the LLM's behavior, purpose, or style, while the user role contains the specific request or question from the user. For example, a system message could tell the model to act as a financial decision-support assistant, while the user message could ask it to evaluate a customer's loan application.

2. A token is a small unit of text that an LLM processes. Words, parts of words, and punctuation can each form tokens. API providers charge based on tokens because longer inputs and outputs require more computational resources than shorter ones, so token usage is a better measure of the work performed than simply counting requests.


```python
#part 1.2
question = "Suggest a name for a savings product for market traders in Accra."

print("====TEMPERATURE 0.0====")

for i in range(5):
  response = ask_llm(question, temperature=0.0, max_tokens=100)
  print(f"Response {i + 1}:")
  print(response.choices[0].message.content)

print("====TEMPERATURE 1.2====")

for i in range(5):
  response = ask_llm(question, temperature=1.2, max_tokens=100)
  print(f"Response {i + 1}:")
  print(response.choices[0].message.content)
```

**Temperature**
1. At temperature 0.0, the responses were generally more consistent and similar to one another. At temperature 1.2, the responses showed greater variation and creativity. For a loan decision-support system, a low temperature such as 0.0-0.3 would be more appropriate because the system should produce consistent and predictable responses rather than highly random recommendations. The final loan decision should also be based on defined financial criteria rather than creativity.
