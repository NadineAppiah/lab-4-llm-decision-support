````python
#part 0
import os
from google.colab import userdata
from openai import OpenAI

API_KEY = userdata.get('GROQ_API_KEY')
if not API_KEY:
  raise ValueError("GROQ_API_KEY not found in environment variables")
print('API key loaded successfully.')

client = OpenAI(
    api_key=API_KEY,
    base_url="https://api.groq.com/openai/v1"
)

MODEL = "llama-3.3-70b-versatile"
print("client ready")
````

````python
#part 1.1
def ask_llm(
    user_prompt,
    system_prompt="You are a helpful assistant.",
    temperature=0.7,
    max_tokens=500
):
    response = client.chat.completions.create(
        model=MODEL,
        messages=[
            {"role": "system", "content": system_prompt},
            {"role": "user", "content": user_prompt}
        ],
        temperature=temperature,
        max_tokens=max_tokens
    )

    return response.choices[0].message.content
print(ask_llm("What is artificial intelligence?"))
````

**Anatomy of a call**

1. The system role provides instructions that define the LLM's behavior, purpose, or style, while the user role contains the specific request or question from the user. For example, a system message could tell the model to act as a financial decision-support assistant, while the user message could ask it to evaluate a customer's loan application.

2. A token is a small unit of text that an LLM processes. Words, parts of words, and punctuation can each form tokens. API providers charge based on tokens because longer inputs and outputs require more computational resources than shorter ones, so token usage is a better measure of the work performed than simply counting requests.

**Temperature**

1. At temperature 0.0, the responses were generally more consistent and similar to one another. At temperature 1.2, the responses showed greater variation and creativity. For a loan decision-support system, a low temperature such as 0.0-0.3 would be more appropriate because the system should produce consistent and predictable responses rather than highly random recommendations. The final loan decision should also be based on defined financial criteria rather than creativity.

**Summarisation prompt**

1. V1 produced a general summary but did not explicitly instruct the model to focus on information important to a loan officer. V2 was more focused because it required the summary to include the applicant's business, loan purpose, financial information, repayment information, and collateral or guarantor information. V2 was also more consistent in length because it explicitly required 3-4 sentences and instructed the model to remain factual and neutral.
2. "No invented details" is essential because the system must base its output only on information actually provided by the applicant. Inventing financial information, collateral, income, or other facts could mislead the loan officer and affect an important financial decision. This type of failure is commonly called an LLM hallucination.

````python
#part 3.2
import json
import pandas as pd

EXTRACT_PROMPT = """
You are a data extraction assistant for a microfinance institution.

Extract information from the loan application letter and return only
a valid JSON object.

The JSON object must contain EXACTLY these keys:
{
    "applicant_name": "string",
    "amount_ghs": number,
    "purpose": "string",
    "monthly_profit_ghs": number or null,
    "has_collateral_or_guarantor": boolean,
    "repayment_months": number or null}

Rules:
1. Extract information only from the letter.
2. If a field is not stated in the letter, use null.
3. Do not invent, assume, or infer information.
4. amount_ghs must be a number, not a string.
5. monthly_profit_ghs must be a number or null.
6. has_collateral_or_guarantor must be true or false.
7. repayment_months must be a number or null.
8. Return ONLY the JSON object.
9. Do not include explanations or introductions.
10. Do not use Markdown code fences.
11. The first character of your response must be { and the last
    character must be }.

Worked example:
Loan Application Letter:
"Dear Manager,
My name is Ama Addo. I run a small fruit stall in Accra.
I am requesting GHS 6,000 to purchase a refrigerator and more stock.
My average monthly profit is GHS 700. My brother will guarantee the loan.
I will repay the loan over 12 months."

Correct JSON:
{
    "applicant_name": "Ama Addo",
    "amount_ghs": 6000,
    "purpose": "refrigerator and more stock",
    "monthly_profit_ghs": 700,
    "has_collateral_or_guarantor": true,
    "repayment_months": 12}
"""

EXTRACT_USER_PROMPT = """
Extract the required fields from this loan application:
{letter_text}
"""

def extract_fields(letter_text, temperature=0):

    response = ask_llm(
        f"{EXTRACT_USER_PROMPT}\n{letter_text}",
        system_prompt=EXTRACT_PROMPT,
        temperature=temperature,
        max_tokens=300
    )

    result = response.choices[0].message.content.strip()

    if result.startswith("```json"):
        result = result[7:]

    if result.startswith("```"):
        result = result[3:]

    if result.endswith("```"):
        result = result[:-3]

    result = result.strip()

    start = result.find("{")
    end = result.rfind("}")

    if start != -1 and end != -1:
        result = result[start:end + 1]

    try:
        return json.loads(result)

    except json.JSONDecodeError:
        print(f"Error decoding JSON: {result}")
        return None
results = []

for letter_id, letter_text in LETTERS.items():
    extracted = extract_fields(letter_text)
    if extracted is not None:
        extracted["letter_id"] = letter_id
        results.append(extracted)

df = pd.DataFrame(results)
df
````

**Structured extraction**

1. The few-shot example should not come from the six letters because using one of the evaluation letters could make the prompt specifically tailored to that data. This could give the model information about the expected answers and make the evaluation less representative of how well the extraction system generalizes to unseen applications.
2. "Use null, do not guess" prevents the model from filling missing information with assumptions. Without this instruction, an LLM may generate plausible-looking values that were never provided in the letter. This is especially problematic in loan applications because invented financial information could mislead the loan officer.
3. Temperature 0 is appropriate for extraction because we want consistent and predictable outputs rather than creative variations. The task is to accurately extract information that already exists in the letter, not to generate new content. Higher temperatures are more appropriate for tasks where creativity and variety are desired.

````python
#part 3.3
BRIEF_PROMPT = """
You are an assistant supporting a microfinance loan officer.

Your role is to organize information and highlight factors that a human
loan officer should consider. You must NOT make the final loan decision.

Using the loan application letter and the extracted information, produce
a concise decision-support brief with exactly these four sections:

1. Strengths
- List positive factors that are explicitly supported by the letter.
- Use bullet points.

2. Risks / Red Flags
- List potential concerns or risks supported by the letter.
- Use bullet points.
- Do not invent risks that are not supported by the information provided.

3. Missing Information
- List important information or documents that the loan officer should
  request before making a decision.
- Use bullet points.

4. Suggested Next Step
- Suggest an appropriate action for the loan officer, such as:
  "request documents", "invite for interview", or "flag for senior review".
- Do NOT say "approve" or "reject".

Important:
- Base the brief only on the information provided.
- Do not invent facts.
- Do not make assumptions about the applicant.
- The final lending decision must be made by a human loan officer.
- Keep the brief concise and professional.

"""

def generate_brief(letter_text, extracted):
    user_prompt = f"""
    Loan Application Letter:
    {letter_text}

    Extracted Information:
    {json.dumps(extracted, indent=2)}
    """

    response = ask_llm(
        user_prompt,
        system_prompt=BRIEF_PROMPT,
        temperature=0,
        max_tokens=300
    )

    return response

briefs = {}
for letter_id, letter_text in LETTERS.items():
    extracted = extract_fields(letter_text)
    if extracted is not None:
      brief = generate_brief(letter_text, extracted)
      briefs[letter_id] = brief

for letter_id in ["L001","L002", "L006"]:
  print(f"Letter {letter_id}:")
  print(briefs[letter_id])
````

````python
#part 4.2
def reliability_test(letter_id, temperature, runs=5):
    results = []
    for i in range(runs):
        result = extract_fields(LETTERS[letter_id])
        results.append(result)

    valid_results = [r for r in results if r is not None]
    result_strings = [
        json.dumps(r, sort_keys=True)
        for r in valid_results
    ]
    unique_results = set(result_strings)

    return {
        "temperature": temperature,
        "valid_json": len(valid_results),
        "identical_values": len(unique_results) == 1,
        "unique_outputs": len(unique_results),
        "results": results
    }
results_temp0 = []

for i in range(5):
    result = extract_fields(LETTERS["L004"], temperature=0)
    results_temp0.append(result)

results_temp1 = []
for i in range(5):
    result = extract_fields(LETTERS["L004"], temperature=1.0)
    results_temp1.append(result)

valid_temp0 = [r for r in results_temp0 if r is not None]
valid_temp1 = [r for r in results_temp1 if r is not None]

unique_temp0 = set(
    json.dumps(r, sort_keys=True)
    for r in valid_temp0
)
unique_temp1 = set(
    json.dumps(r, sort_keys=True)
    for r in valid_temp1
)
print("Temperature 0.0")
print(f"Valid JSON: {len(valid_temp0)}/5")
print(f"Unique outputs: {len(unique_temp0)}")
print(f"Identical across runs: {len(unique_temp0) == 1}")

print("\nTemperature 1.0")
print(f"Valid JSON: {len(valid_temp1)}/5")
print(f"Unique outputs: {len(unique_temp1)}")
print(f"Identical across runs: {len(unique_temp1) == 1}")
````

**Appropriateness**
1. Applicants who write poorly in English or have limited formal education could be unfairly harmed. A good business could appear less credible simply because the applicant's letter is poorly written or lacks formal language. Applicants who use local expressions or have less formal documentation could also be disadvantaged. Before deployment, the institution should check where the data is processed and stored, whether the provider retains or uses the data, who can access it, how the data is encrypted, and whether cross-border data transfers comply with applicable Ghanaian data-protection requirements. The institution should also have appropriate agreements and policies governing the handling of customer data.
3. i. Mandatory human review:
The LLM should only provide summaries, extracted information, risks, and suggested next steps. A qualified loan officer must verify the information and make the final lending decision.
ii. Continuous monitoring and audit logging:
The system should securely record its outputs and errors so that the institution can monitor accuracy, hallucinations, and potential unfair patterns over time. Regular reviews should be conducted to identify and correct problems.

For example, changing temperature affects how consistent the model is, while changing the prompt affects what information the model focuses on and how it responds. The difference is that hyperparameters control the model's generation behavior, whereas prompt engineering changes the instructions and context given to the model.
